```python
import pandas as pd
```

```python
def colected():
    df1 = pd.read_excel("03. Rasgos Galeras.xlsx")
    print(len(df1)) #78
    df2 = pd.read_excel("04. Rasgos_SelvaViva.xlsx")
    print(len(df2)) #48
    df_combined = pd.concat([df1, df2], ignore_index=True)
    df_final = df_combined[["Site","Plot","TreeID_2025"]]
    #df_final.to_excel("iii.xlsx")
    return df_final

df = colected()

def add_oldID_JH():
    df1 = pd.read_excel("Galeras_herbario.xlsx") #base completa
    df2 = pd.read_excel("01. Base_SelvaViva.xlsx") 
    df_combined = pd.concat([df1, df2], ignore_index=True)
    oldID = df_combined[["PlotID","treeID","family","genus","species","JH herbarium collections","leaf sample tree", "Colecct_2024","new ID 2024"]]
    oldID['JH herbarium collections'] =oldID['JH herbarium collections'].astype("string")
    oldID[['JH 1', 'JH 2']] = oldID['JH herbarium collections'].str.split(',', expand=True)
    df_combined.to_excel("basesmerged.xlsx")
    print(oldID)
    return oldID

old_ID = add_oldID_JH()

```

```python
df2 = pd.read_excel("Registro-QCA.xlsx")
df2["numeroCampo"] = df2["numeroCampo"].astype("string")
print(df2)

merged = pd.merge(df,old_ID, how="left",left_on=["Plot","TreeID_2025"], right_on= ["PlotID","new ID 2024"])
merged.to_excel("try.xlsx")
print(merged)
merged2 = pd.merge(merged, df2,how="left",left_on=["JH herbarium collections"], right_on= ["numeroCampo"])
merged2.to_excel("test1.xlsx")
```

```python
df = pd.read_excel("Collected_species_GAL_SEV_2025.xlsx")
df["JH herbarium collections"] = df["JH herbarium collections"].astype("string")
print(len(df))
df1 = pd.read_excel("SelvaViva_herbarium.xlsx")
df1 = df1.rename(columns={
    "JH": "collection JH",
    "dupl": "duplicates"
    # "description" stays the same, so no need to rename
})
df1 = df1[["collection JH","duplicates","description","herbaria"]]
df2 = pd.read_excel("Galeras_herbarium.xlsx")
df2 = df2[["collection JH","duplicates","description","herbaria"]]

#unir los herbarios
hrb_final = pd.concat([df1, df2], ignore_index=True)
hrb_final["collection JH"] = hrb_final["collection JH"].astype("string")

merged = pd.merge(df,hrb_final, how="left", left_on= "JH herbarium collections",right_on="collection JH" )
merged.to_excel("Collections_2025.xlsx")
print(len(merged))
```

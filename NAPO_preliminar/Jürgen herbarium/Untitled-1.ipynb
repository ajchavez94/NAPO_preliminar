```python
def base_herbario():
    #df1 = pd.read_excel("Galeras_herbario.xlsx")
    df2 = pd.read_excel("Galeras_herbarium.xlsx")
    df3 = pd.read_excel("SelvaViva_herbarium.xlsx")
    #QCA = pd.read_excel("Registro-QCA.xlsx")
    all = pd.concat([df2, df3], ignore_index=True)

    return all 
```
